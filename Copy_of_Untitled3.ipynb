```python
import math

def prime_or_not(num):
  if num < 2:
    print("Not a prime number")
    return
  if num == 2:
    print("prime number")
    return
  for i in range(2, int(math.sqrt(num))):
    if num % i == 0:
      print("Not a prime number")
      return

  print("prime number")
num = int(input("Enter a number: "))
prime_or_not(num)
```

```python
num_1 = int(input("Enter number: "))
num_2 = int(input("Enter number: "))
num = int(input("How many operations you need to perform?: "))
for i in range(num):
  operation =input("Enter operation to be performed:\n add\n sub\n mul\n div\n mod\n")
  if operation.lower() == "add":
    sum = num_1 + num_2
    print(f"The sum of {num_1} and {num_2} is",sum)
  if operation.lower() == "sub":
    sub = num_1 - num_2
    print(f"The subtration of {num_1} and {num_2} is",sub)
  if operation.lower() == "mul":
    mul = num_1 * num_2
    print(f"The multiplication of {num_1} and {num_2} is",mul)
  if operation.lower() == "div":
    op = input("Float?\n int?\n")
    if num_2 == 0:
      print("Division is not possible with ZERO!")
    else:
      if op == "float":
        div = num_1/num_2
        print(f"The division of {num_1} and {num_2} is",div)
      else:
        div = num_1//num_2
        print(f"The division of {num_1} and {num_2} is",div)
  if operation.lower() == "mod":
    if num_2 == 0:
      print("Division is not possible with ZERO!")
    else:
      mod = num_1%num_2
      print(f"The remainder for {num_1} and {num_2} is",mod)
  if num_1 and num_2 == 0:
    print("Enter any numbers")
```
